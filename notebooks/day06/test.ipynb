```python
import asyncio
import selectors
import nest_asyncio
from langgraph.checkpoint.postgres.aio import AsyncPostgresSaver
from app.core.config import settings

# 必须加这行，允许在 Jupyter 里用 asyncio.run
nest_asyncio.apply()

async def init_db():
    async with AsyncPostgresSaver.from_conn_string(settings.db.checkpoint_url) as checkpointer:
        await checkpointer.setup()
        print("✅ 完美！建表成功！")

# 强行指定使用 SelectorEventLoop
asyncio.run(init_db(), loop_factory=lambda: asyncio.SelectorEventLoop(selectors.SelectSelector()))
```
