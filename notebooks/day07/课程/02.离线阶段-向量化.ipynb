```python
from langchain_community.embeddings import  DashScopeEmbeddings
import numpy as np
embedding = DashScopeEmbeddings(model = "text-embedding-v4")
# 1. 单条文本向量化,使用embed_query()方法
query = "我爱加班啊"
query_vector = embedding.embed_query(query)
print(f"问题:{query}")
print(f"问题的向量:维度{len(query_vector)},前五维度:{query_vector[:5]}")

# 2. 多条文本的批量向量化
doc_list = ["我爱加班","我要躺平","拒绝躺平"]
doc_vector_list = embedding.embed_documents(doc_list)
print(f"文档列表:{doc_list}")

print(f"向量数量{len(doc_vector_list)},第一个向量的维度{len(doc_vector_list[0])}")

# 3. 计算相似度
for i, doc_vector in enumerate(doc_vector_list):
    # 计算query_vector向量和text_vector的点积
    dot = np.dot(query_vector, doc_vector)
    # 计算两个向量的余弦相似度： 点积 / （向量1长度 ★ 向量2长度）
    cosine = dot / (np.linalg.norm(query_vector) * np.linalg.norm(doc_vector))#输出结果
    print(f"用户问题：{query} 和 {doc_list[i]} 的余弦相似度为{cosine}")
```
