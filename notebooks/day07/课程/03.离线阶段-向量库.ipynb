```python
from langchain_community import embeddings
from langchain_community.embeddings import DashScopeEmbeddings
from langchain_core.vectorstores import InMemoryVectorStore
from langchain_text_splitters import MarkdownHeaderTextSplitter

#准备向量模型
embeddings = DashScopeEmbeddings(model="text-embedding-v4")
#加载文档
with open("../resources/评估.md","r",encoding='utf-8')as f:
    md_doc = f.read()

#文本切分
splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[("#","h1"),("##","h2"),("###","h3")],
    strip_headers=False,
)
doc_list = splitter.split_text(md_doc)

#给每个文档设置id，方便crud
for i,doc in enumerate(doc_list):
    doc.id = f"doc_{i}"
    print(doc.id,end=" | ")
#存储向量库
#   1.创建向量库对象
vector_store = InMemoryVectorStore(embedding=embeddings)
#   2.文档存入数据库
ids = vector_store.add_documents(doc_list)
#   3.根据id删某个文档
vector_store.delete(ids = ['doc_0','doc_1'])
#   4.检索，根据用户问题，从向量库找相似的前k条文档片段
query = "教育的作用是什么"
result_doc_list = vector_store.similarity_search(query)
for result_doc in result_doc_list:
    print("=======================================")
    print(result_doc.model_dump_json())
```
