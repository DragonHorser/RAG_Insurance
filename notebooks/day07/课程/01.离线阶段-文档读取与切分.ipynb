```python
from mineru import MinerU
# MinerU加载PDF文件示例 --轻量解析不需要消耗token
#1.创建客户端对象，获取相关配置信息
client = MinerU()
#2. 解决pdf文件。其中 ocr：光学字符识别技术，即从图片中读取文字的技术
result = client.flash_extract(source="../resources/贵州茅台研报.pdf", is_ocr=True)
print(result.markdown)
```

```python
from app.core.config import settings
from mineru import MinerU
#MinerU加载PDF文件示例--精准解析，需要token
client = MinerU(token=settings.rag.mineru_token)

result = client.extract(source="../resources/small_ocr.pdf", ocr=True)

result.save_markdown(path="../result.md", with_images=True)
```

```python
#结构化切分md格式
from langchain_text_splitters import MarkdownHeaderTextSplitter
with open("../resources/评估.md","r",encoding="utf-8")as f:
    md_text = f.read()

splitter = MarkdownHeaderTextSplitter(
    headers_to_split_on=[("#","h1"),("##","h2"),("###","h3")]
)
n = "/n"
down = "==================================================================================================================================="
doc_list = splitter.split_text(md_text)
for doc in doc_list:
    print("")
    print(f"{down}{n}{doc.metadata}{n}{doc.page_content}")
```
