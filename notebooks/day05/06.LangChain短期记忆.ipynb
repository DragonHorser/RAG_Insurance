```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import InMemorySaver
#加载配置文件
load_dotenv()
#初始化模型
model = init_chat_model("deepseek-flash",extra_body = {"thinking":{"type":"disabled"}})
#创建智能体
agent = create_agent(model = model,tools = [],checkpointer=InMemorySaver())
#定义会话所属
Config = {"configurable":{"thread_id":"t1"}}
#第一次对话
response = agent.invoke(
    {"messages":[HumanMessage(content = "我是练习时长两年的练习生蔡徐坤，喜欢唱跳Rap和篮球")]},
    config = Config
)
print("==============第一次会话=================")
print(response["messages"][-1].content)
#第二次对话
response2 = agent.invoke(
    {"messages":[HumanMessage(content = "我是谁，我喜欢干嘛你还记得不？")]},
    config = Config
)
print("==============第二次会话=================")
print(response2["messages"][-1].content)
#获取会话
print("==============获取会话记忆=================")
agent_status = agent.get_state(Config)
print(agent_status)
print("==============删掉记忆后再次获取会话记忆=================")
agent.checkpointer.delete_thread("t1")
agent_status = agent.get_state(Config)
print(agent_status)



```
