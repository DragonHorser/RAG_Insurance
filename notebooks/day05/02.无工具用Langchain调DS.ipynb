```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model


#加载配置，临时存储到系统变量中去
load_dotenv()

#初始化模型,会自动读取环境变量中的"DEEPSEEK_API_KEY"变量
ds_model = init_chat_model(model="deepseek-flash",extra_body={"thinking":{"type":"disabled"}})

#调用模型
response = ds_model.invoke("你是谁，能干嘛？")
print(response)
```

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage,HumanMessage

load_dotenv()
model = init_chat_model(model = "deepseek-flash",extra_body = {"thinking":{"type":"disabled"}})

response = model.invoke([SystemMessage(content="你是一个现代浪漫主义色彩浓厚的诗人"),
                        HumanMessage(content ="写一首表白诗歌，不需要太长，200字以内即可")
                         ])
print(response)
```
