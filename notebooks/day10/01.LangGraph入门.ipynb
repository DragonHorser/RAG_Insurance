```python
from typing import TypedDict

from langgraph.graph import StateGraph
from langgraph.constants import START,END


#一、先定义State。整个图流转过程中的数据类型(类)
class SimpleState(TypedDict):
    name:str
    greeting:str

#二、在定义Node。整个图需要多少个节点，就定义多少个节点(函数)
def greet_node(state: SimpleState):
    print(f"greet节点收到数据：{state}")
    #不需要亲自更新state数据，只要返回“想更新的数据”即可
    return{"greeting":f"hello,{state["name"]}"}

def uppercase_node(state:SimpleState):
    print(f"upper节点收到数据:{state}")
    greet_upper = state["greeting"].upper()
    return {"greeting":greet_upper}
#三、把这些节点结合Edge，构建成图Graph
builder = StateGraph(SimpleState)
builder.add_node('greet',greet_node)
builder.add_node('uppercase',uppercase_node)

builder.add_edge(START,'greet')
builder.add_edge('greet','uppercase')
builder.add_edge('uppercase',END)

graph = builder.compile()
#四、调用Graph,讲初始数据交给图(数据必须符合State的规定)，允许结果将讲State的最终数据输出
res = graph.invoke(input = {'name':'tom'})
print(res)
```

```python
from typing import TypedDict

from langgraph.graph import StateGraph
from langgraph.constants import START,END

class SimpleState(TypedDict):
    name:str
    greeting:str

def greet_node(state:SimpleState):
    print(f"greet节点收到数据:{state}")
    greeting = f"hello,{state["name"]}"
    return{"greeting":greeting}

def uppercase_node(state:SimpleState):
    print(f"upper几点收到数据:{state}")
    greet_upper = state['greeting'].upper()
    return {"greeting":greet_upper}

builder = StateGraph(SimpleState)
builder.add_node("greet",greet_node)
builder.add_node('uppercase',uppercase_node)

builder.add_edge(START,'greet')
builder.add_edge('greet','uppercase')
builder.add_edge('uppercase',END)

graph = builder.compile()

res = graph.invoke(input = {'name':"tomi"})
print(res)
```
